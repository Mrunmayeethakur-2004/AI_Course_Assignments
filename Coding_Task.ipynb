# Leap Year Logic

```python
def is_leap_year(year):
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

# Test
print(is_leap_year(2024))  # True
print(is_leap_year(1900))  # False
print(is_leap_year(2000))  # True
```

# Identity vs Equality **Explanation:**
- `L1 == L2` checks *equality* — whether the two lists have the same contents/values. Since both lists contain [1, 2, 3], this is True.
- `L1 is L2` checks *identity* — whether both variables point to the same object in memory. Even though their contents are equal, L1 and L2 are two separate list objects, so this is False.

```python
L1 = [1, 2, 3]
L2 = [1, 2, 3]

print(L1 == L2)  # True
print(L1 is L2)  # False
```

# Prime Number Finder

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

count = 0
num = 2
primes = []
while count < 10:
    if is_prime(num):
        primes.append(num)
        count += 1
    num += 1

print(primes)
```

# Nested Conditionals

```python
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "Fail"

print(grade(95))  # A
print(grade(72))  # C
print(grade(50))  # Fail
```

# Reverse an Integer

```python
def reverse_integer(n):
    reversed_num = 0
    n_abs = abs(n)
    while n_abs > 0:
        digit = n_abs % 10
        reversed_num = reversed_num * 10 + digit
        n_abs //= 10
    return -reversed_num if n < 0 else reversed_num

print(reverse_integer(12345))  # 54321
print(reverse_integer(-6789))  # -9876
```

# Runner-Up Score

```python
def runner_up(scores):
    highest = second_highest = float('-inf')
    for score in scores:
        if score > highest:
            second_highest = highest
            highest = score
        elif highest > score > second_highest:
            second_highest = score
    return second_highest

print(runner_up([10, 20, 20, 30, 5]))  # 20
```

# Character Frequency Counter

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("hello world"))
```
